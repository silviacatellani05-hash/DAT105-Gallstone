Now we implement our first Modell (logistic regression) to obtain a baseline:

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedKFold, cross_val_score

X,y = df.drop(columns=["Gallstone Status"]), df["Gallstone Status"]

#Standard Scalar weil Werte unterschiedlich groß sind und so normiert werden
#logistic Regression ein Klassifikationsverfahren
modell = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
#Teilt das Datenset in 5 Teile ein
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=5) #random_state is basically a seed

#eigentliches Trainieren und schauen, wie gut unser Modell abschneidet
ergebnis = cross_val_score(modell, X, y, cv=cv, scoring="roc_auc")
# Ergbnis gibt die Wahrscheinlichkeiten von den 5 Durchgängen aus und berechnet dann mean
print(ergebnis, ergebnis.mean())
```

Second Modell: Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=300, random_state=5) # same random state to make values comparable

ergebnis = cross_val_score(rf, X, y, cv=cv, scoring="roc_auc")
print(ergebnis, ergebnis.mean())
```

Third Modell: Boosting Algorithm

```python
from sklearn.ensemble import GradientBoostingClassifier

gb = GradientBoostingClassifier(
    n_estimators=200, learning_rate=0.05, max_depth=3, random_state=5
)

ergebnis = cross_val_score(gb, X, y, cv=cv, scoring="roc_auc")
print(ergebnis, ergebnis.mean())
```

```python
from xgboost import XGBClassifier
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=5)

xgb = XGBClassifier(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=3,
    subsample=0.8,
    eval_metric="logloss",
    random_state=42,
)

ergebnis = cross_val_score(xgb, X, y, cv=cv, scoring="roc_auc")
print(ergebnis, ergebnis.mean())
```

=> XGB classifier is the best on this data set
